## 5. Why ML beats a fixed rule here

A fixed rule such as:

"If CTR is below 1%, then the content is declining"

is too simplistic because content performance depends on many factors at the same time.

Examples include:

- Click-through rate
- Search impressions
- Average search position
- Engagement rate
- Content age
- Content type
- User behavior

These variables interact in complex ways that are difficult to capture with manually written rules.

Machine learning can learn these patterns automatically from historical data and provide more accurate predictions than fixed thresholds.

The output of the model supports editors and content managers by helping them prioritize which content should be reviewed first.

The prediction supports content editors by helping them prioritize which pages should be reviewed or refreshed first. Instead of manually checking thousands of content items, the model highlights those that are most likely to require attention.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

comparison = df[
    [
        "clicks_90d",
        "impressions_90d",
        "ctr",
        "engagement_rate",
        "trend_direction",
    ]
].copy()

# Simple rule:
comparison["rule_prediction"] = comparison["ctr"] < 1

display(comparison.head(10))

print("""
Observation

This simple rule only considers CTR.

However, content performance depends on multiple interacting signals,
including impressions, clicks, engagement rate, content type,
content age, search position, and many other variables.

Machine Learning can learn these complex relationships and produce
better predictions than a manually defined threshold.
""")
```
