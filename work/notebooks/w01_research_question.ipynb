## 2. The question: decision, action, cost of a wrong call

**Research Question**

Which web pages are most likely to experience declining search traffic in the near future?

**Decision**

Help an SEO team decide which pages should be reviewed first.

**Action**

Review the selected pages, update outdated content, improve internal links, and optimize SEO before traffic declines further.

**Cost of a Wrong Recommendation**

If the model incorrectly flags healthy pages, the team wastes time and resources. If the model misses pages that are actually declining, valuable search traffic may continue to decrease before action is taken.

**Why Machine Learning**

Large websites can contain thousands of pages. Reviewing every page manually is expensive and time-consuming. Machine learning can rank pages by risk so that people focus on the pages that are most likely to need attention.
